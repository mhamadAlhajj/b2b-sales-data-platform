## DM-602 · Build the semantic model
**Type:** Design · **Priority:** High · **Estimate:** 1 day

**What to do**
- Load the dimensions and facts you need from `dwh`. Do not flatten them into one wide table.
- Define the relationships (cardinality, direction, active or inactive) so that they match your star schema.

- Hide technical columns, set data types, formats, sort orders, and hierarchies where they help users.
- Draw or export the model view and compare it with your star schema ERD.
